## Ejercicio 2

1. Escribe un programa que pida al usuario un número y determine si es positivo, negativo o cero.

```python
print("################### Bienvenido usuario ###################")
numero = float(input("Ingresa un número: ")) # Se pide al usuario un número y se transforma a flotante asi se considera tanto decimales como enteros.
if numero > 0: # Condición mayor a 0
  print(f"El número {numero} es positivo")
elif numero < 0: # Condición menos a 0
  print(f"El número {numero} es negativo")
else: # Si no cumplen los anteriores entonces es 0
  print(f"El número {numero} es cero")
```

2. Utiliza un ciclo `for` para imprimir los números del 1 al 10. Luego, utiliza un ciclo `while` para hacer lo mismo.

```python
for num in range(1, 11): # Por cada número en un rango del 1 al 11 (min valor, max valor), el último valor no se incluye.
  print(num) # Imprime el número
```

```python
i = 1
while i < 11: # Mientras el valor i sea menor a 11 se ejecutará. También puede ser como condición i <= 10.
  print(i) # Se imprime el valor
  i += 1 # Se va sumando +1 al valor antes de reiniciar el ciclo
```

## Ejercicio 3

1. Define una cadena de texto y aplica los siguientes métodos: `upper()`, `lower()`,`strip()`, `split()`, y `replace()`. Imprime los resultados.

```python
cadenaDeTexto = " texto 1 y TEXTO 2 "
upper = cadenaDeTexto.upper() # Entrega texto en mayúscula
lower = cadenaDeTexto.lower() # Entrega texto en minúsculas
strip = cadenaDeTexto.strip() # Elimina los espacios al inicio y al final de un texto
split = cadenaDeTexto.split("y") # Crea una lista donde la separación se define en el caracter "y", también puede ser comas o espacios.
replace = cadenaDeTexto.replace("y", "o") # Reemplaza palabras o caracteres de una cadena de texto, en este caso se reemplaza "y" por "o"

print(f"""
Las cadenas de texto se muestran así con los siguientes métodos:
upper = {upper}
lower = {lower}
strip = {strip}
split = {split}
replace = {replace}
""")
```

2. Define una lista con al menos 5 elementos y aplica los siguientes métodos: `append()`,`extend()`, `insert()`, `remove()`, y `pop()`. Imprime los resultados después de cada operación.

```python
lista = [1, 2, 3, 4, 5] # Se define lista con 5 elementos enteros
print("Lista original: ", lista)

lista.append(6) # Se agrega un elemento nuevo
print("Lista con append: ",lista)

lista.extend(["elemento7", "elemento8", "elemento9"]) # Se agrega varios elementos nuevos, se hace una extensión para otra lista
print("Lista con extend: ",lista)

lista.insert(3, "elemento7") # Se inserta en la posición 3 el elemento definido, en este caso "elemento7"
print("Lista con insert: ", lista)

lista.remove("elemento7") # Se remueve el primer elemento que encuentra que es igual al elemento descrito
print("Lista con remove: ", lista)

pop = lista.pop() # Se obtiene el elemento a eliminar en este caso el último. Se puede especificar con la posición
print("Elemento eliminado: ",pop) # Se imprime el elemento a eliminar
print("Lista con pop: ",lista) # Se imprime la nueva lista sin ese elemento
```

2. Define una función que reciba una lista de números y retorne el mayor de ellos.

```python
def numeroMayor(secuencia):
  return max(secuencia) # Se aplica función max para obtener el maximo valor de una secuencia

def numeroMayorCiclo(lista):
  nMayor = lista[0] # El número mayor se define inicialmente es el primer elemento de la lista
  for n in lista: # Por cada elemento en la lista
    if n > nMayor: # Si el elemento actual es mayor al número mayor entonces ese elemento es el nuevo nMayor
      nMayor = n
  return nMayor # Una vez terminado el ciclo completo se retorna el valor final de nMayor

secuencia = [3,5,6,8,3]
print("Número mayor con función max:", numeroMayor(secuencia))
print("Número mayor con ciclo for:", numeroMayorCiclo(secuencia))
```

3. Define una función que reciba una cadena de texto y retorne el número de palabras que contiene.

```python
def contandoPalabras(cadena):
  listaCadena = cadena.strip().split(" ") # Se convierte en lista y cada elemento se define mediante separación de espacios
  return len(listaCadena) # La cantidad de palabras es igual al largo de la lista de cadenas

cadena = "Hola Mundo soy un texto de ejemplo "
print("Cantidad de palabras:", contandoPalabras(cadena))
```
